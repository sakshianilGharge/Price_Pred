```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv('data.csv')
```

```python
df.info()
```

```python
df['year']=pd.to_numeric(df['year'], errors='coerce')
df.dropna(subset=['year'], inplace=True)
df['year']=df['year'].astype(int)
```

```python
df['Price'].unique()
```

```python
for i in df.index:
    df.loc[i, 'Price'] = df.loc[i, 'Price'].replace(',', '')

df['Price'] = pd.to_numeric(df['Price'], errors='coerce')
df.dropna(subset=['Price'], inplace=True)
df['Price'] = df['Price'].astype(int)
```

```python
for i in df.index:
    df.loc[i, 'kms_driven'] = df.loc[i, 'kms_driven'].replace(',', '')
    df.loc[i, 'kms_driven'] = df.loc[i, 'kms_driven'].replace(' kms', '')   
df['kms_driven'] = pd.to_numeric(df['kms_driven'], errors='coerce')
df.dropna(subset=['kms_driven'], inplace=True)
df['kms_driven'] = df['kms_driven'].astype(int)
```

```python
df=df[df['fuel_type']!='LPG']
df.dropna(subset=['fuel_type'], inplace=True)
#df['fuel_type'].unique()
```

```python
for i in df.index:
    name=df.loc[i, 'name']
    names=name.split()
    df.loc[i, 'name']=' '.join(names[1:3])
df['name'].unique()
```
