```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
df=pd.read_csv('cleaned_data_pred.csv')
```

```python
df.info()
```

```python
data=df['company'].value_counts()
plt.figure(figsize=(12, 9))
plt.bar(data.index, data.values, color='#c4a321')
plt.xticks(rotation=90)
plt.xlabel('Companies', color='navy', fontsize=16)
plt.ylabel('Count of Companies', color='navy', fontsize=16)
plt.title("Company-wise Count of vcehicles", color='navy', fontsize=16)
for x, y, in zip(data.index, data.values):
    plt.text(x, y, y, ha='center', va='bottom', fontsize=6)
plt.show()
```

```python
data=df['year'].value_counts().sort_index()
ax = data.plot(kind='bar', figsize=(12, 7), color='#4C72B0')
ax.bar_label(ax.containers[0], fontsize=7, fontweight='bold')
plt.xlabel('Year', color='#1F3A93', fontsize=16)
plt.ylabel('Count', color='#1F3A93', fontsize=16)
plt.title("Year-wise Count of vcehicles", color='#1F3A93', fontsize=16)
plt.show()
```

```python
data=df['fuel_type'].value_counts()
data.plot(kind='pie', autopct='%1.1f%%', figsize=(6, 6), startangle=90)
plt.title('Fuel type Distribution')
plt.ylabel('')
plt.show()
```

```python
min_kms=df['kms_driven'].min()
max_kms=df['kms_driven'].max()
print("Min Kms:", min_kms)
print("Max Kms:", max_kms)
```

```python
plt.figure(figsize=(10,5))
df=df[(df['kms_driven']>=1000)&(df['kms_driven']<=200000)]
sns.swarmplot(x='company', y='kms_driven', data=df)
plt.xticks(rotation=90)
plt.show()
```
